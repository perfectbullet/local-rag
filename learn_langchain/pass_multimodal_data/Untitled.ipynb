```python
from langchain_core.messages import HumanMessage
from langchain_ollama import ChatOllama
```

```python
model = ChatOllama(model="llava", openai_api_key="ollama", openai_api_base="http://localhost:11434/")
```

```python
import base64

# ConnectTimeout: _ssl.c:990: The handshake operation timed out

# image_data = base64.b64encode(httpx.get(image_url).content).decode("utf-8")


# import base64
with open("universe.png", "rb") as imagefile:
    convert = base64.b64encode(imagefile.read())
image_data = convert.decode('utf-8')
```

```python
message = HumanMessage(
    content=[
        {"type": "text", "text": "图片描绘了什么"},
        {
            "type": "image_url",
            "image_url": {"url": f"data:image/jpeg;base64,{image_data}"},
        },
    ],
)
response = model.invoke([message])
print(response.content)
```
